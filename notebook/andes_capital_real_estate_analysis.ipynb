# Andes Capital Real Estate Analysis

Interactive real estate business analysis developed in **Tableau**, focused on executive performance, commercial drivers, customer segmentation, time trends, and cohort behavior.

**Tableau Public:** https://public.tableau.com/views/AndesCapitalRealEstateDashboard/AndesCapitalRealEstateAnalysis?:language=es-ES&:sid=&:redirect=auth&:display_count=n&:origin=viz_share_link


## Business Questions

- What is the company's overall commercial performance?
- Which property types, cities, sales channels, and customer segments generate the most revenue?
- How has revenue evolved over time and year over year?
- Do customers purchase again after their first transaction?


## Key Metrics

- **Total Revenue:** $6.01B
- **Total Sales:** 8,500
- **Average Ticket:** $707.4K
- **Total Commission:** $200.63M
- **YoY Growth:** 11.14%


## Dashboard Structure

1. **Executive Overview** — KPI cards, monthly revenue trend, city comparison, and YoY growth.
2. **Commercial Analysis** — property type, sales channel, customer segment, and a conditional-formatting performance table.
3. **Customer Cohort Analysis** — cohort month versus sale month to evaluate repeat-purchase behavior.


## Tableau Techniques Used

- Calculated fields and LOD expressions
- KPI design and Year-over-Year analysis
- Interactive filters and dashboard actions
- Percentage-of-total calculations and tooltips
- Heatmaps and conditional formatting
- Customer segmentation and cohort analysis
- Tableau Story and data storytelling


## Key Findings

- **Houses** generated the highest total revenue; **Commercial** properties had the highest average ticket; **Apartments** led sales volume.
- **Mexico City** generated more revenue than Bogotá.
- The **Broker** channel accounted for approximately **73% of total revenue**.
- **First-Time Buyers** represented approximately **63% of total revenue**.
- Revenue showed recurring peaks around **March–April** and **September–November**.
- Initial cohort sales declined from **224 in Jan 2023** to **26 in Dec 2024**.
- Repeat purchases remained present across cohorts, with periodic rebounds around the same seasonal windows.


## Strategic Recommendations

- Protect Broker-channel performance while testing opportunities to increase Direct-channel contribution.
- Prioritize First-Time Buyer acquisition and investigate the decline in initial sales among newer cohorts.
- Use Houses for revenue-driving strategies, Commercial properties for higher-ticket opportunities, and Apartments for volume-oriented strategies.
- Align campaigns and sales activity with recurring demand peaks around March–April and September–November.
